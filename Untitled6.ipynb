```python
beta = 0.5

total_precision = 0
total_recall = 0
total_fbeta = 0

for _, row in train_source1.iterrows():

    s1_id = row["entity_id"]
    block_key = row["block_key"]

    # Exact candidates from S2 + S3
    candidates = (
        s2_mapping.get(block_key, set()) |
        s3_mapping.get(block_key, set()) 
    )

    # Ground truth
    actual = ground_truth_dict.get(s1_id, set())

    tp = len(candidates & actual)

    if len(candidates) == 0:
        precision = 1.0 if len(actual) == 0 else 0.0
    else:
        precision = tp / len(candidates)

    if len(actual) == 0:
        recall = 1.0 if len(candidates) == 0 else 0.0
    else:
        recall = tp / len(actual)

    if precision + recall == 0:
        fbeta = 0
    else:
        fbeta = (
            (1 + beta**2) * precision * recall
            / (beta**2 * precision + recall)
        )

    total_precision += precision
    total_recall += recall
    total_fbeta += fbeta

n = len(train_source1)

print("Baseline Precision:", total_precision / n)
print("Baseline Recall:", total_recall / n)
print("Baseline F0.5:", total_fbeta / n)
```

```python
exact_correct = 0
exact_total = 0

for _, row in train_source1.iterrows():

    s1_id = row["entity_id"]
    block_key = row["block_key"]

    candidates = (
        s2_mapping.get(block_key, set()) |
        s3_mapping.get(block_key, set())
    )

    actual = ground_truth_dict.get(s1_id, set())

    if len(candidates) > 0:
        exact_total += 1

        if len(candidates & actual) > 0:
            exact_correct += 1

print("S1 with exact candidates:", exact_total)
print("S1 with at least one correct exact candidate:", exact_correct)
print(
    "Candidate recall:",
    exact_correct / exact_total
)
```

```python
def calculate_match_score(s1_name, s2_name, s1_address, s2_address):

    # Name similarity
    name_matrix = vectorizer.transform([
        s1_name,
        s2_name
    ])

    name_score = cosine_similarity(
        name_matrix[0:1],
        name_matrix[1:2]
    )[0][0]

    # Address similarity
    address_matrix = vectorizer.transform([
        s1_address,
        s2_address
    ])

    address_score = cosine_similarity(
        address_matrix[0:1],
        address_matrix[1:2]
    )[0][0]

    # Combined score
    final_score = (
        0.6 * name_score +
        0.4 * address_score
    )

    return name_score, address_score, final_score


print("Scoring function ready")
```

```python
s1_row = train_source1.iloc[0]
s2_row = candidate_s2.iloc[0]

name_score, address_score, final_score = calculate_match_score(
    s1_row["name_normalized"],
    s2_row["name_normalized"],
    s1_row["address_normalized"],
    s2_row["address_normalized"]
)

print("Source1:", s1_row["entity_id"])
print("Source2:", s2_row["entity_id"])
print("Name score:", name_score)
print("Address score:", address_score)
print("Final score:", final_score)
```

```python
threshold = 0.75

tp = 0
predicted_total = 0
actual_total = 0

for i in range(100):

    s1 = train_source1.iloc[i]

    actual = ground_truth_dict.get(
        s1["entity_id"], set()
    )

    candidates = (
        s2_mapping.get(s1["block_key"], set()) |
        s3_mapping.get(s1["block_key"], set())
    )

    predicted = set()

    for candidate_id in candidates:

        if candidate_id.startswith("S2-"):
            source = train_source2
        else:
            source = train_source3

        candidate = source[
            source["entity_id"] == candidate_id
        ]

        if len(candidate) == 0:
            continue

        candidate = candidate.iloc[0]

        _, _, final_score = calculate_match_score(
            s1["name_normalized"],
            candidate["name_normalized"],
            s1["address_normalized"],
            candidate["address_normalized"]
        )

        if final_score >= threshold:
            predicted.add(candidate_id)

    tp += len(predicted & actual)
    predicted_total += len(predicted)
    actual_total += len(actual)

precision = tp / predicted_total if predicted_total else 0
recall = tp / actual_total if actual_total else 0

f05 = (
    1.25 * precision * recall /
    (0.25 * precision + recall)
    if precision + recall > 0 else 0
)

print("Threshold:", threshold)
print("True Positives:", tp)
print("Predicted:", predicted_total)
print("Actual:", actual_total)
print("Precision:", precision)
print("Recall:", recall)
print("F0.5:", f05)
print("F0.5 percentage:", f05 * 100)
```

```python
batch = holdout_candidate_pairs[:BATCH_SIZE]

batch_scores = []

for s1_id, candidate_id in batch:

    s1_data = s1_entity_map[s1_id]
    candidate_data = get_candidate_row(candidate_id)

    _, _, final_score = calculate_match_score(
        s1_data["name_normalized"],
        candidate_data["name_normalized"],
        s1_data["address_normalized"],
        candidate_data["address_normalized"]
    )

    batch_scores.append(
        (s1_id, candidate_id, final_score)
    )

print("Batch processed:", len(batch_scores))
print("First 5 scores:")
print(batch_scores[:5])
```

```python
for threshold in thresholds:

    predicted = {
        (s1_id, candidate_id)
        for s1_id, candidate_id, score in batch_scores
        if score >= threshold
    }

    correct = sum(
        1
        for s1_id, candidate_id in predicted
        if candidate_id in ground_truth_dict.get(s1_id, set())
    )

    precision = correct / len(predicted) if predicted else 0

    actual_total = sum(
        len(ground_truth_dict.get(s1_id, set()))
        for s1_id, _, _ in batch_scores
    )

    recall = correct / actual_total if actual_total else 0

    f05 = (
        (1 + 0.5**2) * precision * recall
        / ((0.5**2 * precision) + recall)
        if precision + recall > 0 else 0
    )

    print(
        f"Threshold {threshold}: "
        f"Predicted={len(predicted)}, "
        f"Correct={correct}, "
        f"Precision={precision:.4f}, "
        f"Recall={recall:.4f}, "
        f"F0.5={f05:.4f}"
    )
```

```python
threshold_array = np.array(thresholds)

predicted_counts = np.zeros(len(thresholds), dtype=np.int64)
correct_counts = np.zeros(len(thresholds), dtype=np.int64)

total_batches = (
    len(holdout_candidate_pairs) + BATCH_SIZE - 1
) // BATCH_SIZE

for batch_no in range(total_batches):

    start = batch_no * BATCH_SIZE
    end = min(start + BATCH_SIZE, len(holdout_candidate_pairs))

    batch = holdout_candidate_pairs[start:end]

    s1_ids = []
    candidate_ids = []
    s1_addresses = []
    candidate_addresses = []

    for s1_id, candidate_id in batch:
        s1_ids.append(s1_id)
        candidate_ids.append(candidate_id)

        s1_addresses.append(
            s1_entity_map[s1_id]["address_normalized"]
        )

        candidate_data = get_candidate_row(candidate_id)

        candidate_addresses.append(
            candidate_data["address_normalized"]
        )

    address_matrix_1 = vectorizer.transform(s1_addresses)
    address_matrix_2 = vectorizer.transform(candidate_addresses)

    address_scores = np.asarray(
        address_matrix_1.multiply(address_matrix_2).sum(axis=1)
    ).ravel()

    scores = 0.6 + (0.4 * address_scores)

    for i, threshold in enumerate(threshold_array):

        predicted_mask = scores >= threshold
        predicted_counts[i] += predicted_mask.sum()

        correct = 0

        for j in np.where(predicted_mask)[0]:
            if candidate_ids[j] in ground_truth_dict.get(
                s1_ids[j], set()
            ):
                correct += 1

        correct_counts[i] += correct

    if (batch_no + 1) % 100 == 0 or batch_no == total_batches - 1:
        print(
            f"Processed {batch_no + 1}/{total_batches} batches "
            f"({end}/{len(holdout_candidate_pairs)} pairs)"
        )

print("Full optimized evaluation completed.")
print("Total candidate pairs:", len(holdout_candidate_pairs))
print("Holdout actual matches:", holdout_actual_total)
```
